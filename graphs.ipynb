```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("Car data.csv")

# Create Brand and Model columns from Car Name
df["Brand"] = df["Name"].str.split().str[0]
df["Model"] = df["Name"].str.split().str[1]

# Set manufacturing date range
start_date = pd.to_datetime("2010-01-01")
end_date = pd.to_datetime("2026-01-01")

# Find total days between dates
days = (end_date - start_date).days

# Generate random manufacturing dates
random_days = np.random.randint(0, days, len(df))
df["Manufacturing Date"] = start_date + pd.to_timedelta(random_days, unit="D")

# Calculate car age
today = pd.to_datetime("today")
df["Age Days"] = (today - df["Manufacturing Date"]).dt.days

# Convert days into years
df["Car Age"] = round(df["Age Days"] / 365.25, 2)

# Display data
print(df.head())
```

```python
import numpy as np

# Calculate total income from Onroad Price
total_income = df["Onroad Price"].sum()
print("Total Income:", total_income)


# Select only numerical columns
num_data = df.select_dtypes(include=np.number)

# Convert DataFrame into NumPy array
arr = num_data.to_numpy()

print("Array Shape:", arr.shape)
print("Array Size:", arr.size)


# Find highest Onroad Price
highest_price = np.max(df["Onroad Price"])
print("Highest Onroad Price:", highest_price)


# Find lowest mileage
lowest_mileage = np.min(df["Range (kmpl)"])
print("Lowest Mileage:", lowest_mileage)


# Calculate average engine capacity
avg_engine = np.mean(df["CC"])
print("Average Engine Capacity:", round(avg_engine, 2))
```

```python
import matplotlib.pyplot as plt
# barchart
plt.figure(figsize=(10,5))

plt.bar(df["Name"], df["Onroad Price"], color="skyblue")

plt.title("Onroad Price of Cars")
plt.xlabel("Car Name")
plt.ylabel("Onroad Price")

plt.xticks(rotation=45, ha = "right")

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.bar(df["Name"], df["Range (kmpl)"])

plt.title("Mileage Comparison")
plt.xlabel("Car Name")
plt.ylabel("Mileage (kmpl)")

plt.xticks(rotation=45,ha="right")

plt.show()
```

```python
#Engine Capacity VS Mileage 
plt.scatter(df["CC"],df["Range (kmpl)"])
plt.title("Engine Capacity VS Mileage ")
plt.xlabel("Engine Capacity (CC)")
plt.ylabel("Range (kmpl)")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.hist(df["Ex-Showroom Price"], bins=10, alpha=0.5, label="Ex-Showroom")
plt.hist(df["Onroad Price"], bins=10, alpha=0.5, label="Onroad Price")

plt.title("Ex-Showroom vs Onroad Price")
plt.xlabel("Price")
plt.ylabel("Number of Cars")

plt.legend()
plt.show()
```

```python
#Distribution of Onroad Price 
plt.hist(df["Onroad Price"],bins =5,edgecolor="black")
plt.xlabel("Onroad Price ")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Count the number of cars of each brand
count = df["Brand"].value_counts()

# Labels for pie chart
labels = count.index

# Colors
colors = ['red', 'blue', 'green', 'yellow', 'orange', 'purple', 'pink']

# Plot pie chart
plt.pie(count, labels=labels, colors=colors, autopct='%1.1f%%')

# Title of chart
plt.title("Brand Distribution")

# Make pie chart circular means equal ratio
plt.axis('equal')

# Display chart. for ouput
plt.show()

```
